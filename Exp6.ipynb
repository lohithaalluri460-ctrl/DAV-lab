```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# Step 2: Create Dataset
data = {
    'Maths': [80, 90, 70, 60, 85],
    'Science': [85, 95, 75, 65, 90],
    'English': [70, 80, 60, 55, 75],
    'Computer': [90, 95, 80, 70, 92]
}

df = pd.DataFrame(data)

print("ORIGINAL DATASET:")
print(df)

# Step 3: NORMALIZATION
# Scaling values between 0 and 1

minmax = MinMaxScaler()

normalized_data = minmax.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\nNORMALIZED DATA:")
print(normalized_df)

# Step 4: STANDARDIZATION
# Mean = 0 and Standard Deviation = 1

scaler = StandardScaler()

standardized_data = scaler.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\nSTANDARDIZED DATA:")
print(standardized_df)

# Step 5: PRINCIPAL COMPONENT ANALYSIS (PCA)
# First standardize the data

scaled_data = StandardScaler().fit_transform(df)

# Reduce 4 features into 2 components

pca = PCA(n_components=2)

pca_data = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)

print("\nPCA REDUCED DATA:")
print(pca_df)

# Step 6: Explained Variance Ratio

print("\nEXPLAINED VARIANCE RATIO:")
print(pca.explained_variance_ratio_)

# Step 7: Total Variance Preserved

total_variance = sum(pca.explained_variance_ratio_)

print("\nTOTAL VARIANCE PRESERVED:")
print(total_variance)

print("\nPERCENTAGE OF VARIANCE PRESERVED:")
print(total_variance * 100, "%")

# Step 8: Final Comparison

print("\nNUMBER OF ORIGINAL FEATURES:", df.shape[1])
print("NUMBER OF PCA FEATURES:", pca_df.shape[1])

print("\nFEATURE ENGINEERING COMPLETED SUCCESSFULLY!")

```
